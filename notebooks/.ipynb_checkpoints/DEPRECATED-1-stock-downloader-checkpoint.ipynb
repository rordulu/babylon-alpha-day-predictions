# LSTM Time Series - Stock Price Prediction
## Part 1 - Data Explanatory Analysis
This notebook focuses on examining the raw dataset containing daily historical prices of Google stocks. The goal is to identify a specific timeframe suitable for further analysis and prediction using LSTM.

> **INPUT**: The raw data file of all available stock prices of Google (Alphabet Inc.), as downloaded from its original source. <br/>
> **OUTPUT**: The extracted historical data of the targeted period for analysis, stored in an intermediary location for next steps.

### 1. INITIALIZATION

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
```

### 2. DOWNLOAD MANY STOCKS TO FILE

```python
import yfinance as yf

stocks_to_use = ['GOOGL', 'AAPL', 'MSFT', 'AMZN',
                 'NFLX', 'KO', 'BAC', 'TSLA', 'GS', 
                 'AVGO', 'SPOT', 'AMD', 'PEP', 'DASH', 
                 'JPM', 'IBKR', 'META', 'LLY', 'V',
                 'UNH', 'XOM', 'MA', 'COST', 'PG', 
                 'WMT', 'JNJ', 'CRM', 'ABBV', 'ORCL', 
                 'MRK', 'CVX', 'CSCO', 'WFC',
                 'ACN', 'NOW', 'PEP', 'MCD', 'IBM',
                 'DIS', 'LIN', 'TMO', 'ADBE', 'ABT',
                 'ISRG', 'PM', 'GE', 'INTU', 'CAT', 'QCOM', 'TXN', 'VZ', 'AXP', 'BKNG', 'T'
                ]

def get_raw_data_for(ticker_symbol):

    # Fetch data
    stock_data = yf.Ticker(ticker_symbol)
    
    # Get historical market data
    historical_data = stock_data.history(period="5y")  # Options: 1d, 5d, 1mo, 3mo, 6mo, 1y, 2y, 5y, 10y, ytd, max
    historical_data = historical_data[['Open', 'High', 'Low', 'Close', 'Volume']]
    historical_data['TICKER'] = ticker_symbol
    
    data_file_name = "..//data//raw//" + ticker_symbol + "_stock_price_recent.csv"
    
    historical_data.to_csv(data_file_name)
    print(historical_data)

for ticker in stocks_to_use:
    get_raw_data_for(ticker)
```

```python
import pandas as pd

all_data_file_name = "..//data//raw//ALL_stock_price_recent.csv"
all_data = None
for ticker in stocks_to_use:
    
    data_file_name = "..//data//raw//" + ticker + "_stock_price_recent.csv"
    data = pd.read_csv(data_file_name)
    
    if all_data is None:
        all_data = pd.DataFrame(columns=data.columns, index=data.index)
        all_data = data
    else:
        print(f"adding new stock data")
        all_data = pd.concat([all_data, data])

if all_data is not None:
    all_data.to_csv(all_data_file_name)

                         
```
