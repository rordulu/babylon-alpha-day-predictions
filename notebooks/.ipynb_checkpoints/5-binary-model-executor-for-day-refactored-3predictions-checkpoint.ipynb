### 3-PREDICTION MODEL EXECUTOR FOR SIMPLE_V3
This notebook is updated to handle the new 3-prediction output structure from simple_v3 models.

#### Key Features:
- **Statistical Thresholds**: Uses mean + sigma for each day/stock instead of fixed target values
- **3 Prediction Types**: Close, High, and Low ratios to next-day open price
- **Dynamic Target Calculation**: Target value is calculated as `Target-20D-Mean + Target-20D-Sigma` from the data
- **Proper Ground Truth**: Binary classification based on whether actual ratio exceeded the statistical threshold

#### How It Works:
1. **Model outputs 3 predictions**: [Close-to-Open, High-to-Open, Low-to-Open] ratios
2. **Target mapping**: Automatically selects the correct prediction index based on target name
3. **Target-specific statistical thresholds**: Each target has its own mean/sigma columns:
   - `Target-20D-Mean_0`, `Target-20D-Sigma_0` for Close-to-Open
   - `Target-20D-Mean_1`, `Target-20D-Sigma_1` for High-to-Open  
   - `Target-20D-Mean_2`, `Target-20D-Sigma_2` for Low-to-Open
4. **Threshold calculation**:
   - **Close & High**: `threshold = mean + sigma` (predicting above-average performance)
   - **Low**: `threshold = mean - sigma` (predicting significantly low performance)
5. **Price calculation**:
   - **Close & High**: `Predicted-Price = Open × (1 + threshold)` 
   - **Low**: `Predicted-Price = Open × (1 + threshold)` where threshold is negative, so effectively `Open × (1 - |threshold|)`
6. **Ground truth**: Whether the actual ratio exceeded the target-specific statistical threshold
7. **Model prediction**: Compared against learned threshold to make binary predictions

```python
import numpy as np
import pandas as pd
import joblib
from sklearn.preprocessing import MinMaxScaler
from keras.models import load_model
from pathlib import Path

# Target to output index mapping for simple_v3 models
TARGET_TO_INDEX = {
    "Next-Day-Close-To-Next-Day-Open-Ratio": 0,
    "Next-Day-High-To-Next-Day-Open-Ratio": 1,
    "Next-Day-Low-To-Next-Day-Open-Ratio": 2
}

# Updated combos for simple_v3 models with 3-prediction structure
# Format: [stock, model_version, target, score]
combos = [
    ["MSTR", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["MSTR", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["AAPL", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["AAPL", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["AAPL", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["NVDA", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["NVDA", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 2],
    ["AVGO", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["AVGO", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["AVGO", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["PLTR", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["PLTR", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 2],
    ["PLTR", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["META", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 3],
    ["META", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 2],
    ["META", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["AMZN", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["AMZN", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["ABNB", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 3],
    ["AMD", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 3],
    ["AMD", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
    ["AMD", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["BA", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["BA", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["BABA", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 3],
    ["BAC", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["BAC", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
    ["BAC", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["DASH", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["CRM", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["CRM", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
    ["CRM", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["CSCO", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["CSCO", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["CSCO", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["DIS", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["DIS", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["MU", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["F", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["GOOGL", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["GOOGL", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["INTC", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["INTC", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
    ["IONQ", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["JNJ", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["JPM", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["JPM", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["JPM", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["KO", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["LLY", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["META", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["MSFT", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["NFLX", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 3],
    ["NFLX", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 2],
    ["NFLX", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 1],
    ["PFE", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["PFE", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 2],
    ["PFE", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 2],
    ["QQQ", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["QQQ", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["QQQ", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["RGTI", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 2],
    ["ROKU", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 3],
    ["ROKU", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 3],
    ["ROKU", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["SHOP", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 2],
    ["SHOP", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["SNAP", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["SNAP", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 2],
    ["SNAP", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["SPY", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["SPY", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
    ["SPY", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 3],
    ["UNH", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["UNH", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["V", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["WFC", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["WFC", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
    ["WFC", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["WMT", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 2],
    ["WMT", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["XOM", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 3],
    ["ZM", "simple_v3", "Next-Day-Close-To-Next-Day-Open-Ratio", 4],
    ["ZM", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
    ["ZM", "simple_v3", "Next-Day-Low-To-Next-Day-Open-Ratio", 4],
    ["SPOT", "simple_v3", "Next-Day-High-To-Next-Day-Open-Ratio", 4],
]

# Set target date - you can modify this to run for different dates
target_date = "2025-07-02"

write_header = True
for combo in combos:
    ss = combo[0]  # stock
    mm = combo[1]  # model version
    tt = combo[2]  # target
    score = combo[3]
    
    # Get prediction index from target name using mapping
    prediction_index = TARGET_TO_INDEX.get(tt, 0)
    
    print(f"SS: {ss}, MM: {mm} TT: {tt}, Prediction Index: {prediction_index}")
    
    # Prepare model location and name - for simple_v3 models
    model_location = "../models/" + ss + "/" + mm + "/" + mm + "/"
    model_name = "model.keras"  # simple_v3 uses generic model name
    
    # Load threshold files specific to the target
    threshold_df = pd.read_csv(model_location + tt + "_threshold.csv")
    threshold = threshold_df.iloc[0]["Threshold"]
    
    # Load detailed thresholds for this target
    lvls = pd.read_csv(model_location + tt + "_detailed_threshold.csv")
    # Filter by prediction index if the CSV has that column
    if 'prediction_index' in lvls.columns:
        lvls = lvls[lvls['prediction_index'] == prediction_index]
    
    if len(lvls) > 0:
        threshold_levels = lvls.iloc[(lvls['threshold'] - threshold).abs().idxmin()]
    else:
        # Fallback to default values
        threshold_levels = pd.Series({
            'TPs': 0, 'FPs': 0, 'FNs': 0, 'Precision': 0.0, 
            'Recall': 0.0, 'Random-Guess': 0.0
        })
    
    # Load target value
    target_val_df = pd.read_csv(model_location + tt + "_binary_target_val.csv")
    target_val = target_val_df.iloc[0]["Target-Val"]

    save_location = "../predictions-simple-v3-3predictions/"
    directory = Path(save_location)
    directory.mkdir(parents=True, exist_ok=True)

    print(f"stock: {ss}, target: {tt}, threshold: {threshold}, target_val: {target_val}, pred_index: {prediction_index}")
    
    # Load the model
    try:
        best_model = load_model(model_location + model_name)
        print(f"Successfully loaded model from {model_location + model_name}")
    except Exception as e:
        print(f"Error loading model: {e}")
        continue
    
    # Construct data for the specific date
    try:
        data_frame, real_target = construct_values_for_model(
            ticker_symbols=[ss], 
            underlying_target=["Next-Day-Close-To-Next-Day-Open-Ratio", "Next-Day-High-To-Next-Day-Open-Ratio", "Next-Day-Low-To-Next-Day-Open-Ratio"],
            sequence_size=dm_sequence_size(mm), 
            use_for_last_day_prediction=False, 
            verbose=False, 
            data_interval="2y",
            refresh=True,
            is_binary_prediction=True,
            incl_earnings=True,
            use_fixed_mean=False
        )

        print(data_frame.columns)
        
        print(f"Constructed data frame with shape: {data_frame.shape}")
        
        # Filter to specific date
        data_frame = get_specific_date_data(data_frame, target_date)
        print(f"Filtered data for {target_date}: {len(data_frame)} rows")
        
        if len(data_frame) == 0:
            print(f"No data available for {target_date} for stock {ss}")
            continue
            
    except Exception as e:
        print(f"Error constructing data: {e}")
        continue

    # Check if next day's data is available
    if pd.isna(data_frame.iloc[-1]['Next-Day-Open']):
        print(f"ABORTING.... NEXT DAY'S DATA IS NOT AVAILABLE FOR THE LAST DAY IN YOUR SELECTION...")
        continue
    
    # Prepare inputs for model prediction
    X_test = np.array(data_frame["LstmData"].to_list())
    X_test_ticker = np.array(data_frame["Ticker"].to_list())
    X_test_sector = np.array(data_frame["Sector"].to_list())
    X_test_yesterday = data_frame[lstm_features].to_numpy()
    X_test_days = np.array(data_frame["Date"].to_list())
    y_test = np.array(data_frame["y-value"].to_list())
    
    # Make prediction - this will return 3 values per sample
    y_predict_all = best_model.predict([X_test, X_test_ticker, X_test_sector, X_test_yesterday])
    print(f"Model prediction shape: {y_predict_all.shape}")
    
    # Extract the specific prediction index we want (0, 1, or 2)
    if y_predict_all.shape[1] >= prediction_index + 1:
        y_predict = y_predict_all[:, prediction_index]
    else:
        print(f"Warning: Model output has {y_predict_all.shape[1]} predictions, but requested index {prediction_index}")
        y_predict = y_predict_all[:, 0]  # fallback to first prediction
    
    data_frame['y-predict'] = y_predict
    
    # Inverse normalize predictions
    for ticker in data_frame["Orig_Ticker"].unique():
        ticker_mask = data_frame["Orig_Ticker"] == ticker
        ticker_predictions = data_frame.loc[ticker_mask, 'y-predict'].values.reshape(-1, 1)
        
        try:
            # For binary predictions, we might not need to inverse normalize
            # but keeping the structure for consistency
            if hasattr(pd.DataFrame(ticker_predictions, columns=['y-predict']), 'values'):
                normalized_preds = inverse_normalize_data(
                    pd.DataFrame(ticker_predictions, columns=['y-predict']), 
                    real_target, 
                    ticker
                )
                if normalized_preds is not None:
                    data_frame.loc[ticker_mask, 'y-predict-original'] = normalized_preds.values.flatten()
                else:
                    data_frame.loc[ticker_mask, 'y-predict-original'] = ticker_predictions.flatten()
            else:
                data_frame.loc[ticker_mask, 'y-predict-original'] = ticker_predictions.flatten()
        except Exception as e:
            print(f"Warning: Could not inverse normalize for {ticker}: {e}")
            data_frame.loc[ticker_mask, 'y-predict-original'] = ticker_predictions.flatten()

    # Calculate statistical thresholds for this specific day and stock
    # For simple_v3 models with multiple targets, each target has its own mean/sigma columns
    target_mean_col = f"Target-20D-Mean_{prediction_index}"
    target_sigma_col = f"Target-20D-Sigma_{prediction_index}"

    print("=== DEBUG INFO FOR TROUBLESHOOTING ===")
    print("Available columns:", data_frame.columns.tolist())
    print("Target-20D-Mean type:", type(data_frame.iloc[0]["Target-20D-Mean"]))
    print("Target-20D-Mean value:", data_frame.iloc[0]["Target-20D-Mean"])
    print("Target-20D-Sigma type:", type(data_frame.iloc[0]["Target-20D-Sigma"]))
    print("Target-20D-Sigma value:", data_frame.iloc[0]["Target-20D-Sigma"])
    print("Looking for columns:", target_mean_col, target_sigma_col)
    print("Columns with Target in name:", [col for col in data_frame.columns if 'Target' in col])
    print("Data shape:", data_frame.shape)
    print("Date:", data_frame.iloc[0]["Date"])
    print("use_fixed_mean was:", True)  # from the function call
    print("==================")

    print(data_frame)
    
    # Special handling for Low predictions (mean - sigma instead of mean + sigma)
    if tt == "Next-Day-Low-To-Next-Day-Open-Ratio":
        data_frame["Statistical-Threshold"] = data_frame[target_mean_col] - data_frame[target_sigma_col]
    else:
        data_frame["Statistical-Threshold"] = data_frame[target_mean_col] + data_frame[target_sigma_col]
    data_frame["Target-Val"] = data_frame["Statistical-Threshold"]
    data_frame["Target-Mean"] = data_frame[target_mean_col]
    data_frame["Target-Sigma"] = data_frame[target_sigma_col]
    
    # Add metadata columns
    data_frame["Threshold"] = threshold
    data_frame["Target"] = tt
    data_frame["Prediction-Index"] = prediction_index
    
    # Add threshold level data
    for col in ['TPs', 'FPs', 'FNs', 'Precision', 'Recall', 'Random-Guess']:
        if col in threshold_levels:
            data_frame[col] = threshold_levels[col]
        else:
            data_frame[col] = 0.0
    
    data_frame["Score"] = score
    data_frame["Mode"] = mm
    
    # Calculate predicted prices based on the target being predicted
    if tt == "Next-Day-Close-To-Next-Day-Open-Ratio":
        data_frame["Predicted-Price"] = data_frame["Next-Day-Open"] * (1 + data_frame["Target-Val"])
        data_frame["Actual-Ratio"] = data_frame["Next-Day-Close-To-Next-Day-Open-Ratio"]
    elif tt == "Next-Day-High-To-Next-Day-Open-Ratio":
        data_frame["Predicted-Price"] = data_frame["Next-Day-Open"] * (1 + data_frame["Target-Val"])
        data_frame["Actual-Ratio"] = data_frame["Next-Day-High-To-Next-Day-Open-Ratio"]
    elif tt == "Next-Day-Low-To-Next-Day-Open-Ratio":
        # For Low predictions, Target-Val is negative (mean - sigma), so we use (1 + Target-Val) 
        # which effectively becomes (1 - |Target-Val|) for the predicted low price
        data_frame["Predicted-Price"] = data_frame["Next-Day-Open"] * (1 + data_frame["Target-Val"])
        data_frame["Actual-Ratio"] = data_frame["Next-Day-Low-To-Next-Day-Open-Ratio"]
    
    # Additional ratio columns for analysis
    data_frame["Close-To-Open-Pct"] = data_frame["Next-Day-Close"] / data_frame["Next-Day-Open"]
    data_frame["High-To-Open-Pct"] = data_frame["Next-Day-High-To-Next-Day-Open-Ratio"]
    data_frame["Low-To-Open-Pct"] = data_frame["Next-Day-Low-To-Next-Day-Open-Ratio"]
    
    # Binary ground truth: whether the actual ratio exceeded the statistical threshold
    data_frame["y-value-balanced"] = data_frame["Actual-Ratio"] >= data_frame["Target-Val"]
    
    # Binary predictions based on model threshold
    data_frame['y-predict-binary'] = data_frame['y-predict-original'] >= threshold
    data_frame['y-predict-binary-with-5-percent-grace'] = data_frame['y-predict-original'] >= threshold * 0.95
    
    print(f"PRINTING POSITIVE PREDICTIONS FOR: {ss} WITH TARGET: {tt} (Prediction Index: {prediction_index})")
    positive_predictions = data_frame[data_frame['y-predict-binary'] == 1.0]
    print(positive_predictions)
    
    # Prepare output columns
    output_columns = [
        "Orig_Ticker", "Date", "y-value", "y-value-balanced", "y-predict", "y-predict-original", 
        "y-predict-binary", "y-predict-binary-with-5-percent-grace", "Score", "Mode", "Threshold", 
        "TPs", "FPs", "FNs", "Precision", "Recall", "Random-Guess", "Target", "Target-Val", 
        "Statistical-Threshold", "Target-Mean", "Target-Sigma", "Actual-Ratio", "Prediction-Index", 
        "Next-Day-Open", "Predicted-Price", "Next-Day-Close", "Close-To-Open-Pct", "High-To-Open-Pct", "Low-To-Open-Pct"
    ]
    
    to_write = data_frame[output_columns]
    to_write.to_csv(save_location + "todays-guess.csv", mode='a', index=False, header=write_header)
    
    to_write_only_true = to_write[to_write["y-predict-binary"] == 1.0]
    to_write_only_true.to_csv(save_location + "todays-guess_only_true.csv", mode='a', index=False, header=write_header)
    
    to_write_only_true_grace = to_write[to_write["y-predict-binary-with-5-percent-grace"] == 1.0]
    to_write_only_true_grace.to_csv(save_location + "todays-guess_only_true_5_percent_grace.csv", mode='a', index=False, header=write_header)
    
    write_header = False
    
    print(f"Completed processing for {ss} - {tt} - Prediction Index {prediction_index}\\n")
```

```python
import pandas as pd
import numpy as np

# Load and clean the results
data_frame_results = pd.read_csv(save_location + "todays-guess_only_true.csv")

# Remove duplicate header rows (where Orig_Ticker equals 'Orig_Ticker')
data_frame_results = data_frame_results[data_frame_results['Orig_Ticker'] != 'Orig_Ticker']

# Convert numeric columns that might be strings due to header contamination
numeric_columns = ['Score', 'y-value-specific', 'y-value-balanced', 'y-predict', 'y-predict-original', 
                  'Target-Val', 'Close-To-Open-Pct', 'High-To-Open-Pct', 'Low-To-Open-Pct',
                  'Prediction-Index', 'Next-Day-Open', 'Predicted-Price', 'Next-Day-Close']

for col in numeric_columns:
    if col in data_frame_results.columns:
        data_frame_results[col] = pd.to_numeric(data_frame_results[col], errors='coerce')

# Convert boolean columns
boolean_columns = ['y-value-balanced', 'y-predict-binary', 'y-predict-binary-with-5-percent-grace']
for col in boolean_columns:
    if col in data_frame_results.columns:
        # Convert to boolean, handling string representations
        data_frame_results[col] = data_frame_results[col].map({
            True: True, False: False, 'True': True, 'False': False, 
            1.0: True, 0.0: False, '1.0': True, '0.0': False
        })

print(f"Data loaded successfully. Shape: {data_frame_results.shape}")
print(f"Score column type: {data_frame_results['Score'].dtype}")
print(f"Score range: {data_frame_results['Score'].min()} to {data_frame_results['Score'].max()}")

# Filter for high-scoring predictions - adjust threshold based on actual data
# Your data has scores 1-3, not 1-4, so use Score >= 2 instead of >= 4
filtered = data_frame_results[data_frame_results["Score"] >= 2]
print(f"After Score >= 2 filter: {len(filtered)} records")

filtered = filtered[filtered["y-predict-binary"] == True]
print(f"After y-predict-binary == True filter: {len(filtered)} records")

if len(filtered) == 0:
    print("No records match the filtering criteria.")
    print(f"Available Score values: {sorted(data_frame_results['Score'].dropna().unique())}")
    print("Consider lowering the Score threshold.")
else:
    # Apply loss mitigation strategy
    print("Applying loss mitigation strategy...")
    # If market closes below the negative target value, cap the loss
    filtered.loc[filtered['Close-To-Open-Pct'] < (1 - filtered['Target-Val']), 'Close-To-Open-Pct'] = (1 - filtered['Target-Val'])
    filtered.loc[filtered['Close-To-Open-Pct'] < 0.985, 'Close-To-Open-Pct'] = 0.985

    # Calculate gains and losses - use y-value-specific instead of y-value-balanced (which contains NaN)
    gains = filtered[filtered["y-value-specific"] == 1.0]["Target-Val"]
    num_gains = gains.count()
    total_gains = gains.sum()

    losses = filtered[filtered["y-value-specific"] == 0.0]
    num_loss = len(losses)
    total_loss_uf = losses["Close-To-Open-Pct"] - 1.0
    total_loss = total_loss_uf.sum()

    print(f"\nRaw calculations:")
    print(f"Gains count: {num_gains}, Losses count: {num_loss}")
    print(f"Total gains (raw): {total_gains:.6f}")
    print(f"Total losses (raw): {total_loss:.6f}")

    # Only proceed if we have data
    if num_gains > 0 or num_loss > 0:
        # Apply transaction cost function (example quadratic function)
        gains_adj = gains + 1
        total_loss_uf_adj = total_loss_uf + 1

        # Transaction cost function: 271.6504x² - 519.3991x + 248.6636
        gains_after_costs = (271.6504 * (gains_adj**2)) - (519.3991 * gains_adj) + 248.6636
        losses_after_costs = (271.6504 * (total_loss_uf_adj**2)) - (519.3991 * total_loss_uf_adj) + 248.6636

        sum_gains = gains_after_costs.sum()
        sum_losses = losses_after_costs.sum()

        print(f"\nAnalysis Results:")
        print(f"{num_gains} Gains: {total_gains:.4f}, {num_loss} Losses: {total_loss:.4f}")
        print(f"AVG Gain: {total_gains/num_gains if num_gains > 0 else 0:.4f}, AVG Loss: {total_loss/num_loss if num_loss > 0 else 0:.4f}")
        print(f"Total Gain (after costs): {sum_gains:.4f}, Total Loss (after costs): {sum_losses:.4f}")
        print(f"Started With: {num_loss + num_gains}, Ended with: {sum_gains + sum_losses:.4f}")
        
        total_trades = num_loss + num_gains
        if total_trades > 0:
            print(f"Calculated GAIN: {100 * ((sum_gains + sum_losses) / total_trades - 1):.2f}%")
        else:
            print("No trades to calculate gain from")
    else:
        print("\nNo gains or losses found in the filtered data.")

    # Show distribution by prediction index
    if 'Prediction-Index' in data_frame_results.columns:
        print("\nDistribution by Prediction Index:")
        pred_index_dist = data_frame_results.groupby('Prediction-Index').agg({
            'y-predict-binary': 'sum',
            'Score': 'mean',
            'y-value-specific': lambda x: x.sum() if len(x) > 0 else 0  # Use y-value-specific instead of y-value-balanced
        }).round(3)
        print(pred_index_dist)

    # Show distribution by Target type
    if 'Target' in data_frame_results.columns:
        print("\nDistribution by Target Type:")
        target_dist = data_frame_results.groupby('Target').agg({
            'y-predict-binary': 'sum',
            'Score': 'mean',
            'y-value-specific': lambda x: x.sum() if len(x) > 0 else 0  # Use y-value-specific instead of y-value-balanced
        }).round(3)
        print(target_dist)

    # Show sample of the filtered data
    print(f"\nSample of filtered data (first 5 rows):")
    print(filtered[['Orig_Ticker', 'Date', 'Score', 'Target', 'y-predict-binary', 'y-value-specific', 'Target-Val', 'Close-To-Open-Pct']].head())
```

```python
# ### TRADING SIMULATION
# Simulate day-by-day trading with the predictions
```

```python
# Trading simulation with multiple predictions per stock
try:
    money = 100
    filtered_trading = filtered.sort_values(by='Date', ascending=True)
    filtered_trading = filtered_trading.reset_index(drop=True)
    
    print("Starting Trading Simulation...")
    
    for day in filtered_trading["Date"].unique():
        trades = filtered_trading[filtered_trading["Date"] == day]
        
        # Determine investment per trade (max 4 trades per day)
        per_invest = money / max(4, len(trades))
        invest_result = money - (per_invest * min(len(trades), 4))
        trades = trades.reset_index(drop=True)
        
        # Handle multiple predictions per stock - keep best performing prediction index per stock
        if 'Prediction-Index' in trades.columns:
            # First, keep the highest scoring prediction per ticker and target combination
            trades = trades.loc[trades.groupby(['Orig_Ticker', 'Target'])['Score'].idxmax()]
            
            # Then, remove duplicates by ticker, keeping the highest target value
            trades = trades.loc[trades.groupby(['Orig_Ticker'])['Target-Val'].idxmax()]
            
            # Finally, shuffle and remove any remaining duplicates
            trades = trades.sample(frac=1, random_state=42).drop_duplicates(subset=['Orig_Ticker'])
        else:
            # Fallback to original logic if no prediction index column
            trades = trades.loc[trades.groupby(['Orig_Ticker', 'Target'])['Score'].idxmax()]
            trades = trades.loc[trades.groupby(['Orig_Ticker'])['Score'].idxmax()]
            trades = trades.loc[trades.groupby(['Orig_Ticker'])['Target-Val'].idxmax()]
            trades = trades.sample(frac=1, random_state=42).drop_duplicates(subset=['Orig_Ticker'])
        
        trades = trades.reset_index(drop=True)
        
        # Limit to maximum 4 trades per day
        if len(trades) > 4:
            trades = trades.nlargest(4, 'Score')
        
        # Execute trades
        for index, trade in trades.iterrows():
            if trade["y-value-balanced"] == True:
                # Profitable trade
                target = trade["Target-Val"] + 1
                gain = (271.6504 * (target**2)) - (519.3991 * target) + 248.6636
                invest_result += per_invest * gain
            else:
                # Loss trade
                target = trade["Close-To-Open-Pct"]
                loss = (271.6504 * (target**2)) - (519.3991 * target) + 248.6636
                invest_result += per_invest * loss
        
        trade_stocks = trades['Orig_Ticker'].unique() if len(trades) > 0 else []
        print(f"END OF DAY: {day}, TRADES: {len(trades)}, MONEY: ${invest_result:.2f}, STOCKS: {list(trade_stocks)}")
        money = invest_result
    
    print(f"\\nFinal Result: ${money:.2f}")
    print(f"Total Return: {((money - 100) / 100 * 100):.2f}%")
    
except Exception as e:
    print(f"Error in trading simulation: {e}")
```

```python
# Function to run predictions for a date range
def run_predictions_for_date_range(start_date, end_date, selected_combos=None):
    """
    Run predictions for a range of dates
    
    Args:
        start_date (str): Start date in YYYY-MM-DD format
        end_date (str): End date in YYYY-MM-DD format  
        selected_combos (list): Optional subset of combos to run
    """
    if selected_combos is None:
        selected_combos = combos
    
    # Generate date range
    date_range = pd.date_range(start=start_date, end=end_date, freq='D')
    
    results_summary = []
    
    for current_date in date_range:
        date_str = current_date.strftime('%Y-%m-%d')
        print(f"\\n=== Processing Date: {date_str} ===")
        
        daily_predictions = 0
        daily_positive_predictions = 0
        
        for combo in selected_combos:
            ss, mm, tt, score = combo
            
            # Get prediction index from target name using mapping
            prediction_index = TARGET_TO_INDEX.get(tt, 0)
            
            # Construct data for this specific date
            data_frame, real_target = construct_values_for_model(
                ticker_symbols=[ss], 
                underlying_target=tt,
                sequence_size=dm_sequence_size(mm), 
                use_for_last_day_prediction=False, 
                verbose=False, 
                data_interval="2y",
                refresh=True,  # Don't refresh for each date to speed up
                is_binary_prediction=True,
                incl_earnings=True,
                use_fixed_mean=False
            )
            
            # Filter to specific date
            date_data = get_specific_date_data(data_frame, date_str)
            
            if len(date_data) == 0:
                continue
            
            # Check if next day's data is available
            if pd.isna(date_data.iloc[-1]['Next-Day-Open']):
                continue
            
            # Load model and make prediction
            model_location = "../models/" + ss + "/" + mm + "/" + mm  + "/"
            model_name = "model.keras"
            
            best_model = load_model(model_location + model_name)
            
            # Prepare inputs
            X_test = np.array(date_data["LstmData"].to_list())
            X_test_ticker = np.array(date_data["Ticker"].to_list())
            X_test_sector = np.array(date_data["Sector"].to_list())
            X_test_yesterday = date_data[lstm_features].to_numpy()
            
            # Make prediction
            y_predict_all = best_model.predict([X_test, X_test_ticker, X_test_sector, X_test_yesterday])
            
            # Extract specific prediction index
            if y_predict_all.shape[1] >= prediction_index + 1:
                y_predict = y_predict_all[:, prediction_index]
            else:
                y_predict = y_predict_all[:, 0]
            
            daily_predictions += 1
            
            # Apply threshold (simplified - using default 0.5)
            if y_predict[0] >= 0.5:
                daily_positive_predictions += 1
                print(f"  POSITIVE: {ss} - Index {prediction_index} - Prediction: {y_predict[0]:.3f}")
        
        results_summary.append({
            'Date': date_str,
            'Total_Predictions': daily_predictions,
            'Positive_Predictions': daily_positive_predictions,
            'Positive_Rate': daily_positive_predictions / daily_predictions if daily_predictions > 0 else 0
        })
        
        print(f"  Daily Summary: {daily_positive_predictions}/{daily_predictions} positive predictions")
    
    # Create summary DataFrame
    summary_df = pd.DataFrame(results_summary)
    
    print("\\n=== Date Range Summary ===")
    print(summary_df)
    
    return summary_df

# Example usage:
# Uncomment the lines below to run for a date range
# summary = run_predictions_for_date_range("2025-04-21", "2025-06-24")
# summary.to_csv(save_location + "date_range_summary.csv", index=False)
```
