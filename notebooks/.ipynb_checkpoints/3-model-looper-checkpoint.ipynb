```python
import papermill as pm

# Input and output notebook paths
input_notebook2 = "2-data-preprocessing.ipynb"
output_notebook_template2 = "2-data-preprocessing_{}.ipynb"
input_notebook3 = "3-model-training-model2.ipynb"
output_notebook_template3 = "3-model-training-model2_{}.ipynb"

# stocks_to_use = ['GOOGL', 'AAPL', 'MSFT', 'AMZN', 'NFLX', 'KO', 'BAC', 'TSLA', 'GS', 'AVGO', 'SPOT', 'AMD', 'PEP', 'DASH', 'JPM', 'IBKR']
stocks_to_use = ['META', 'LLY', 'V', 'UNH', 'XOM', 'MA', 'COST', 'PG', 'WMT', 'JNJ', 'CRM', 'ABBV', 'ORCL', 'MRK', 'CVX', 'CSCO', 'WFC']

# Generate parameter_list dynamically
parameter_list2 = [{"TICKER": stock} for stock in stocks_to_use]
parameter_list3 = [{"TICKER": stock} for stock in stocks_to_use]

# Execute the notebook 2 for each set of parameters
for i, params in enumerate(parameter_list2):
    output_notebook = output_notebook_template2.format(stocks_to_use[i])
    pm.execute_notebook(
        input_notebook2, 
        output_notebook, 
        parameters=params,
        log_output=True
    )
    print(f"Executed notebook {output_notebook} with parameters {params}")
```

```python

# Execute the notebook 3 for each set of parameters
for i, params in enumerate(parameter_list3):
    output_notebook = output_notebook_template3.format(stocks_to_use[i])
    pm.execute_notebook(
        input_notebook3, 
        output_notebook, 
        parameters=params,
        log_output=True
    )
    print(f"Executed notebook {output_notebook} with parameters {params}")
```
