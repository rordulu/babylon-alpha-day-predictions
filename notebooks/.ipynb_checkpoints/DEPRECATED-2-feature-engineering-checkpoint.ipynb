# LSTM Time Series - Stock Price Prediction
## Part 2 - Data Preprocessing
This notebook focuses on processing the filtered dataset containing the historical prices of Google stocks over the past five years.

> **INPUT**: Filtered dataset containing Google's stock prices from the last five years, obtained from the previous phase. <br/>
> **OUTPUT**: Preprocessed and transformed data divided into training, validation, and testing subsets, stored in an interim location for the training phase.

### 1. INITIALIZATION

```python
# Import necessary libraries and modules
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt
import joblib
import matplotlib.dates as mdates
import numpy as np
```

### DEFINE THE CONSTANTS FOR THIS NOTEBOOK

```python
# Set output width
pd.set_option("display.width", 120)

stocks_to_use = ['GOOGL', 'AAPL', 'MSFT', 'AMZN',
                 'NFLX', 'KO', 'BAC', 'TSLA', 'GS', 
                 'AVGO', 'SPOT', 'AMD', 'PEP', 'DASH', 
                 'JPM', 'IBKR', 'META', 'LLY', 'V',
                 'UNH', 'XOM', 'MA', 'COST', 'PG', 
                 'WMT', 'JNJ', 'CRM', 'ABBV', 'ORCL', 
                 'MRK', 'CVX', 'CSCO', 'WFC',
                 'ACN', 'NOW', 'PEP', 'MCD', 'IBM',
                 'DIS', 'LIN', 'TMO', 'ADBE', 'ABT',
                 'ISRG', 'PM', 'GE', 'INTU', 'CAT', 'QCOM', 'TXN', 'VZ', 'AXP', 'BKNG', 'T'
                ]

# Define selected features and target attribute
features = ['Date', 'Open', 'High', 'Low', 'Close', 'Volume', 'TICKER', 'High-Low-Ratio', 'High-Open-Ratio', 'Close-Open-Ratio',
       'Next-Day-Open']
            # , '200-D-MA', '100-D-MA', '50-D-MA', '20-D-MA']
features_to_be_used = ["Open", "High", "Low", "Close", "Volume", "High-Low-Ratio", "Close-Open-Ratio", "Next-Day-Open"]
                       # , "Open-To-Prev-Close-Ratio", "Close-To-Prev-Close-Ratio",
                       # "Open-To-Prev-Open-Ratio", "High-To-Prev-High-Ratio", "Low-To-Prev-Low-Ratio", "Volume-To-Prev-Volume-Ratio"]
                       # , "200-D-MA", "100-D-MA", "50-D-MA", "20-D-MA"]
target = "High-Open-Ratio"

SEQUENCE_SIZE = 64
```

### 2. LOADING DATASETS

```python
def get_raw_data_for(ticker_symbol):
    # Fetch data
    data_file_name = "..//data//raw//" + ticker_symbol + "_stock_price_recent.csv"
    data = pd.read_csv(data_file_name)
    print(f"data length moment#1: {len(data)}")
    # Convert date column to a valid Datetime format
    data["Date"] = pd.to_datetime(data["Date"])

    # Feature engineering:
    data["High-Low-Ratio"] = data["High"] / data["Low"]
    data["High-Open-Ratio"] = data["High"] / data["Open"]
    data["Close-Open-Ratio"] = data["Close"] / data["Open"]
    data["Next-Day-Open"] = data['Open'].shift(-1)
    # data['Open-To-Prev-Close-Ratio'] = data['Open'].div(data['Close'].shift())
    # data['Close-To-Prev-Close-Ratio'] = data['Close'].div(data['Close'].shift())
    # data['Open-To-Prev-Open-Ratio'] = data['Open'].div(data['Open'].shift())
    # data['High-To-Prev-High-Ratio'] = data['High'].div(data['High'].shift())
    # data['Low-To-Prev-Low-Ratio'] = data['Low'].div(data['Low'].shift())
    # data['Volume-To-Prev-Volume-Ratio'] = data['Volume'].div(data['Volume'].shift())
    # data['200-D-MA'] = data['Close'].rolling(window=200).mean()
    # data['100-D-MA'] = data['Close'].rolling(window=100).mean()
    # data['50-D-MA'] = data['Close'].rolling(window=50).mean()
    # data['20-D-MA'] = data['Close'].rolling(window=20).mean()
    
    # data['Pos-To-200D-MA'] = data["Close"] / data['Close'].rolling(window=200).mean()
    # data['Pos-To-100D-MA'] = data["Close"] / data['Close'].rolling(window=100).mean()
    # data['Pos-To-50D-MA'] = data["Close"] / data['Close'].rolling(window=50).mean()
    # data['Pos-To-20D-MA'] = data["Close"] / data['Close'].rolling(window=20).mean()
    # data['Pos-To-200D-MA-DoD'] = data['Pos-To-200D-MA'].div(data['Pos-To-200D-MA'].shift())
    # data['Pos-To-100D-MA-DoD'] = data['Pos-To-100D-MA'].div(data['Pos-To-100D-MA'].shift())
    # data['Pos-To-50D-MA-DoD'] = data['Pos-To-50D-MA'].div(data['Pos-To-50D-MA'].shift())
    # data['Pos-To-20D-MA-DoD'] = data['Pos-To-20D-MA'].div(data['Pos-To-20D-MA'].shift())
    # data['Log-Compared-To-Prev-Close-For-Close'] = np.log1p(data['Compared-To-Prev-Close-For-Close'])
    # data['Sq-Close-To-Prev-Close-Ratio'] = data['Close-To-Prev-Close-Ratio'] ** 2
    
    # data['Open-Log'] = np.log(data['Open'])
    # data['Close-Log'] = np.log(data['Close'])
    # data['High-Log'] = np.log(data['High'])
    # data['Low-Log'] = np.log(data['Low'])
    # data['Open-Base'] = data['Open'] / min(data['Open'])
    # data['Close-Base'] = data['Close'] / min(data['Close'])
    # data['High-Base'] = data['High'] / min(data['High'])
    # data['Low-Base'] = data['Low'] / min(data['Low'])

    print(f"data length moment#2: {len(data)}")
    print(f"printing print(data.columns) #1: {data.columns}")
    
    # data.replace([np.inf, -np.inf], np.nan, inplace=True)
    # data = data.dropna()
    
    # Check loaded data shape
    data.shape
    # Check loaded data head
    data.head()
    # Check column types
    data.dtypes
    
    # print(f"DEBUG: {data.shape} + HEAD: {data.head()}")
    
    print(f"data length moment#3: {len(data)}")
    return data


def normalize_data(data, ticker):
    # Initialize scaler with range [0,1]
    sc = MinMaxScaler(feature_range=(0,1))
    
    # Fit and transform scaler to training set
    data[features_to_be_used] = sc.fit_transform(data[features_to_be_used])

    # Prepare scaler model name and location
    scaler_model_location = "..//models//"
    scaler_model_name = ticker + "_stock_price_scaler"
    scaler_model_ext = "gz"

    # Store scaler model
    joblib.dump(sc, scaler_model_location + scaler_model_name + "." + scaler_model_ext)

    return data
```

```python
# Define a method to construct the input data X and Y
def construct_lstm_data(raw_data, ready_data, ticker):
    """
    Construct input data (X) and target data (y) for LSTM model from a pandas DataFrame.

    Parameters:
    -----------
    data : numpy.ndarray
        Input data array of shape (n_samples, n_features).
    
    sequence_size : int
        Number of previous time steps to use as input features for predicting the next time step.
    
    target_attr_idx : int
        Index of column in `data` DataFrame that corresponds to target attribute that LSTM model will predict.

    Returns:
    --------
    data_X : numpy.ndarray
        Array of LSTM input sequences of shape (n_samples - sequence_size, sequence_size, n_features).

    data_y : numpy.ndarray
        Corresponding target values for each input sequence of shape (n_samples - sequence_size,).
    """

    # print(f"inside construct_lstm_data with shape: {data.shape}")
    
    # Initialize constructed data variables
    data_X = []
    data_y = []
    target_attr_idx = features.index(target)

    if (len(ready_data) != len(raw_data)):
        raise ExceptionType("Please control the row and ready data you provided for lstm data construction") 
    data_length = len(ready_data)
    print(f"data length moment#4: {len(raw_data)}")
    print(f"data length moment#4.5: {len(ready_data)}")
    
    # Iterate over the dataset
    for i in range(SEQUENCE_SIZE, data_length):
        new_data = ready_data.iloc[i-SEQUENCE_SIZE:i,0:ready_data.shape[1]]
        
        # Columns to scale
        columns_to_scale = ["Open", "Close", "High", "Low"]

        # print(f"new_data[columns_to_scale] OLD HEAD : {new_data[columns_to_scale].head()} OLD TAIL: {new_data[columns_to_scale].tail()} size: {len(new_data)}")
        
        # Scale only the selected columns
        new_data[columns_to_scale] = new_data[columns_to_scale].apply(lambda col: col / col.max(), axis=0)

        # print(f"new_data[columns_to_scale] NEW : {new_data[columns_to_scale].head()} OLD TAIL: {new_data[columns_to_scale].tail()} size: {len(new_data)}")

        data_X.append(new_data)
        # Calculate if next day we had > 1% increase
        up_by_1_percent = 1.0 if raw_data.iloc[i, target_attr_idx] >= 1.010 else 0.0
        if up_by_1_percent == 1.0:
            print(f"newly found y-value=1 for ticker: {ticker} at index: {i}, high-open-ratio at: {raw_data.iloc[i, target_attr_idx]}")
        else:
            print(f"could not find y-value=1 for ticker: {ticker} at index: {i}, high-open-ratio at: {raw_data.iloc[i, target_attr_idx]}")
        data_y.append(up_by_1_percent)
            
    # Return constructed variables
    return np.array(data_X), np.array(data_y)
```

### GET RAW DATA AND COMBINE PREV 64 DAYS data for LSTM

```python
X = np.empty((0, SEQUENCE_SIZE, len(features_to_be_used)))
y = np.empty((0))
dates = np.empty(0)
tickers = np.empty(0)

for ticker in stocks_to_use:
    ticker_x_raw = get_raw_data_for(ticker)
    # ticker_x_raw = ticker_x_raw[-510:] # Last ~ 2 YEARS
    ticker_x_raw_normalized = normalize_data(ticker_x_raw, ticker)
    ml_model_ready_ticker_x = ticker_x_raw_normalized[features_to_be_used]
    print(ml_model_ready_ticker_x.head())
    ticker_x, ticker_y = construct_lstm_data(ticker_x_raw, ml_model_ready_ticker_x, ticker)
    ticker_dates = ticker_x_raw["Date"][SEQUENCE_SIZE:len(ticker_x_raw)]

    X = np.concatenate((X, ticker_x), axis=0)
    y = np.concatenate((y, ticker_y), axis=0)
    dates = np.concatenate((dates, ticker_dates), axis=0)
    tickers = np.concatenate((tickers, np.full(ticker_dates.size, ticker, dtype=object)))

    # print(ticker_y)
    
    print(f"ticker_x shape: {ticker_x.shape}")
    print(f"X shape: {X.shape}")
    print(f"ticker_y shape: {ticker_y.shape}")
    print(f"y shape: {y.shape}")
    print(f"ticker_dates shape: {ticker_dates.shape}")
    print(f"dates shape: {dates.shape}")
    print(f"tickers shape: {tickers.shape}")

# for i in range(len(X)):
#     print(f"Stock: {tickers[i]}, Date: {dates[i]}, y-value: {y[i]}")
```

```python
# Prepare datasets files and location
data_file_location = "../data/processed/"
data_file_X = "X_stock_price_processed_train"
data_file_y = "y_stock_price_processed_train"
data_file_dates = "dates_stock_price_processed_train"
data_file_tickers = "tickers_stock_price_processed_train"
data_file_ext = "npy"

np.save(data_file_location + data_file_X + "." + data_file_ext, X)
np.save(data_file_location + data_file_y + "." + data_file_ext, y)
np.save(data_file_location + data_file_dates + "." + data_file_ext, dates)
np.save(data_file_location + data_file_tickers + "." + data_file_ext, tickers)

print(f"ALL COMPLETED")
```
