#### Building LSTM Model

```python
class CustomBatchNormalization(tf.keras.layers.Layer):
    def __init__(self, **kwargs):
        super(CustomBatchNormalization, self).__init__(**kwargs)

    def call(self, inputs):
        # Compute batch mean and variance
        mean = tf.reduce_mean(inputs, axis=0)
        variance = tf.math.reduce_variance(inputs, axis=0)
        epsilon = 1e-6  # Small constant to prevent division by zero

        # Normalize inputs
        normalized = (inputs - mean) / tf.sqrt(variance + epsilon)

        # Adjust the first element to be 1.0
        first_element_scale = 1.0 / (normalized[0] + epsilon)  # Ensure first element becomes 1.0
        normalized = normalized * first_element_scale

        return normalized
```

```python
class AppendParametersLayer(tf.keras.layers.Layer):
    def call(self, inputs):
        inner_layer_output, extra_params = inputs
        return tf.concat([inner_layer_output, extra_params], axis=-1)
```

```python
### MODEL 1
from tensorflow.keras.layers import Concatenate, ZeroPadding1D, Cropping1D, Embedding, RepeatVector

input_layer = Input(shape=(X_train.shape[1], X_train.shape[2]), name="values")
stock_id_input = Input(shape=(1,), name="stock_id")  # Stock ID input

# Stock Embedding Layer
num_stocks = 100  # Adjust based on the number of unique stocks
embedding_dim = 4  # Embedding dimension
stock_embedding = Embedding(input_dim=num_stocks, output_dim=embedding_dim, name="stock_embedding")(stock_id_input)
# Reshape to remove the sequence length dimension
stock_embedding = Reshape((embedding_dim,))(stock_embedding)
# expanded_embedding = RepeatVector(X_train.shape[1])(stock_embedding)  # Match timesteps with input data

# Concatenate stock embeddings with input features
# combined_input = Concatenate(axis=-1)([input_layer, expanded_embedding])

cnn = Conv1D(128,3)(input_layer)
cnn = PReLU(shared_axes=[1], alpha_initializer=tf.initializers.constant(0.25))(cnn)
cnn = Conv1D(128,3)(cnn)
cnn = PReLU(shared_axes=[1], alpha_initializer=tf.initializers.constant(0.25))(cnn)
cnn = Conv1D(256,3)(cnn)
cnn = PReLU(shared_axes=[1], alpha_initializer=tf.initializers.constant(0.25))(cnn)
cnn = Conv1D(256,3)(cnn)
cnn = PReLU(shared_axes=[1], alpha_initializer=tf.initializers.constant(0.25))(cnn)
cnn = Conv1D(512,3)(cnn)
cnn = PReLU(shared_axes=[1], alpha_initializer=tf.initializers.constant(0.25))(cnn)

expanded_embedding = RepeatVector(cnn.shape[1])(stock_embedding)  # Match timesteps with input data
cnn = Concatenate(axis=-1)([cnn, expanded_embedding])

lstm = LSTM(units = 800, return_sequences = True)(cnn)
lstm = Dropout(rate = 0.2)(lstm)
lstm = LSTM(units = 400, return_sequences = True)(lstm)
lstm = Dropout(rate = 0.2)(lstm)
lstm = LSTM(units = 200, return_sequences = True)(lstm)
lstm = Dropout(rate = 0.2)(lstm)
lstm = LSTM(units = 200)(lstm)
lstm = Dropout(rate = 0.2)(lstm)
lstm = Dense(1,activation='sigmoid')(lstm)
```

```python
regressor = Model([stock_id_input,input_layer],lstm)
regressor.summary()
```

### Compile the model

```python
optimizer = Adam(learning_rate=STARTING_LR)
loss = None
loss_name = None

if LOSS_FUNCTION == "mse":
    loss = "mse"
    loss_name = "mse"
elif LOSS_FUNCTION == "focal":
    loss = focal_loss(alpha=0.25, gamma=2.0)
    loss_name = "focal_loss"
else:
    loss = "binary_crossentropy"
    loss_name = "binary_crossentropy"

regressor.compile(
    optimizer=optimizer,
    loss=loss,
    metrics=[
        precision_with_threshold(0.7),  # Custom precision
        recall_with_threshold(0.7),    # Custom recall
        'AUC',                         # Built-in AUC
        'Precision',                   # Built-in Precision
        'Recall',                      # Built-in Recall
        f1_score_metric,                # Custom F1 score
        true_positives,
        all_positives,
        recall_mul_prediction
    ]
)
```

### Adjust parameters for the training

```python
file_name_variable = f"special_{SPEC_FILE_TAG}_epoch_{NUM_EPOCH}_lr{STARTING_LR}_batch{BATCH_SIZE}_seqsize{SEQUENCE_SIZE}_loss{loss_name}"

# Create a checkpoint to monitor the validation loss and save the model with the best performance.
model_location = "..//models//"
model_name = f"model_{file_name_variable}.keras"
best_model_checkpoint_callback = ModelCheckpoint(
    model_location + model_name, 
    monitor="val_AUC", 
    save_best_only=True, 
    mode="max", 
    verbose=0)

# Create the ReduceLROnPlateau callback
reduce_lr = ReduceLROnPlateau(
    monitor='loss',     # Metric to monitor
    factor=0.333,             # Factor to reduce the learning rate (new_lr = lr * factor)
    patience=REDUCE_LR_PATIENCE,             # Number of epochs with no improvement before reducing
    min_lr=1e-11             # Minimum learning rate
)

# EarlyStopping to stop training
early_stopping = EarlyStopping(
    monitor='val_loss', patience=STOP_L_PATIENCE, restore_best_weights=True
)

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train),
    y=y_train
)
class_weights = dict(enumerate(class_weights))
```

```python
import logging

# Configure logging
logging.basicConfig(
    filename=f"training_output_{file_name_variable}.log", 
    level=logging.INFO, 
    format='%(asctime)s - %(message)s'
)
logger = logging.getLogger()

# Define a custom callback to log training progress
class LogCallback(tf.keras.callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        logger.info(f"Epoch {epoch + 1}: {logs}")
```

### START THE TRAINING

```python
import keras

keras.config.disable_traceback_filtering()

X_train_tickers = y_date_ticker_train[:, 2].astype(X_train.dtype).reshape(-1, 1)
X_validate_tickers = y_date_ticker_validate[:, 2].astype(X_validate.dtype).reshape(-1, 1)

# print(f"X_train type: {type(X_train)}")
# print(f"X_train_tickers type: {type(X_train_tickers)}")

# print("X_train shape:", X_train.shape)
# print("X_train_tickers shape:", X_train_tickers.shape)

# x_train_dict = {
#     "values": X_train,  # Input1's data
#     "stock_id": X_train_tickers   # Input2's data
# }

# x_validate_dict = {
#     "values": X_validate,  # Input1's data
#     "stock_id": X_validate_tickers   # Input2's data
# }

# Training the model
history = regressor.fit(
    [X_train_tickers, X_train]
    ,y_train 
    ,validation_data=([X_validate_tickers, X_validate], y_validate)
    ,epochs=NUM_EPOCH
    ,batch_size = BATCH_SIZE
    ,callbacks = [best_model_checkpoint_callback, reduce_lr, early_stopping, LogCallback()]
    # ,class_weight=class_weights
)
```

#### Performance Evaluation

```python
# Visualizing model performance during training
plt.figure(figsize=(18, 6))

plt.plot(history.history["loss"][1:], label="Training Loss")
plt.plot(history.history["val_loss"][1:], label="Validation Loss")

plt.title("LSTM Model Performance")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

The chart above highlights the following observations:
- Training loss values decrease over the epochs, starting from a higher value and gradually decreasing. This indicates that the model is learning to fit the training data better as training progresses.
- Similarly, the validation loss values also decrease initially, indicating improvement in the model's ability to generalize to unseen data (validation set).
- Some fluctuations in the validation loss have been observed, which is quite justifiable due to the high variations and complexity in the price performance over time.
- There are no signs of overfitting since both loss values are continuously decreasing along with each other over the epochs.
- Validation loss has maintained lower values than training loss across almost the whole training period, which is quite normal due to the complexity/size of training data.
