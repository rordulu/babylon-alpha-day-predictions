# XGBoost (On top of ARIMA)- Stock Price Prediction
## Part 4 - Model

### 1. INITIALIZATION

```python
# Import necessary libraries and modules
import matplotlib.dates as mdates
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, LSTM, Dropout, Dense
import joblib
from keras.callbacks import ModelCheckpoint
from keras.models import load_model
import statsmodels.api as sm
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.tsa.arima.model import ARIMA
from sklearn import metrics
from statsmodels.tsa.stattools import adfuller
```

### 2. LOADING DATASETS

```python
# Prepare data file location and name
data_file_location = "..//data//processed//"
data_file_name_train = "google_stock_price_processed_train"
data_file_name_validate = "google_stock_price_processed_validate"
data_file_name_test = "google_stock_price_processed_test"
data_file_ext = "csv"

# Load data files
data_train_df = pd.read_csv(data_file_location + data_file_name_train + "." + data_file_ext)
data_validate_df = pd.read_csv(data_file_location + data_file_name_validate + "." + data_file_ext)
data_test_df = pd.read_csv(data_file_location + data_file_name_test + "." + data_file_ext)
```

```python
# Check loaded datasets shape
print(f"Training Dataset Shape: {data_train_df.shape}")
print(f"Validation Dataset Shape: {data_validate_df.shape}")
print(f"Testing Dataset Shape: {data_test_df.shape}")
```

```python
# Display a summary of each dataset
print("Training Dataset:")
print(data_train_df.head())
print("Validation Dataset:")
print(data_validate_df.head())
print("Testing Dataset:")
print(data_test_df.head())
```

### 3. DATA PREPROCESSING

The dataset has already undergone the primary preprocessing and transformation in the pervious phase. This section is only to display the data and extract features and dates.

```python
# Convert Date column to a valid Datetime format
data_train_df["Date"] = pd.to_datetime(data_train_df["Date"])
data_validate_df["Date"] = pd.to_datetime(data_validate_df["Date"])
data_test_df["Date"] = pd.to_datetime(data_test_df["Date"])
```

```python
# Extract dates from each dataset
data_train_dates = data_train_df["Date"]
data_validate_dates = data_validate_df["Date"]
data_test_dates = data_test_df["Date"]
```

```python
# Extract features
features = ["Open", "High", "Low",	"Close", "Adj Close", "Volume"]
data_train_scaled = data_train_df[features].values
data_validate_scaled = data_validate_df[features].values
data_test_scaled = data_test_df[features].values
```

```python
# HELPMER METHODS
def make_plot(figsize=(10, 6), title=None, xlabel=None, ylabel=None):
    plt.figure(figsize=figsize)
    if title:
        plt.title(title)
    if xlabel:
        plt.xlabel(xlabel)
    if ylabel:
        plt.ylabel(ylabel)

def adf_test(temp):
    # p-value>0.562 or Critical Value(1%)>-3.44, non-stationary
    t = adfuller(temp)
    output = pd.DataFrame(index=['Test Statistic Value', 'p-value', 'Lags Used', 'Number of Observations Used', 'Critical Value(1%)', 'Critical Value(5%)', 'Critical Value(10%)'], columns=['value'])
    output['value']['Test Statistic Value'] = t[0]
    output['value']['p-value'] = t[1]
    output['value']['Lags Used'] = t[2]
    output['value']['Number of Observations Used'] = t[3]
    output['value']['Critical Value(1%)'] = t[4]['1%']
    output['value']['Critical Value(5%)'] = t[4]['5%']
    output['value']['Critical Value(10%)'] = t[4]['10%']
    print(output)
```

Let's do some ARIMA

```python
#ARIMA
#Feature engineering with DIFF

# Step 4: Feature engineering - First and second-order differences
data_train_df['diff_1'] = data_train_df['Close'].diff(1)
data_train_df['diff_2'] = data_train_df['diff_1'].diff(1)

# Plot first-order difference
make_plot(title='First-Order Difference', xlabel='Time', ylabel='First Difference')
data_train_df['diff_1'].plot(title='')
plt.show()

# Plot second-order difference
make_plot(title='Second-Order Difference', xlabel='Time', ylabel='Second Difference')
data_train_df['diff_2'].plot(title='Second-Order Difference')
plt.show()
```

```python
# ARIMA
# White noise test
# Step 5: White noise test
print(acorr_ljungbox(data_train_df['diff_1'].dropna(), lags=2, boxpierce=True))

# Step 6: Auto-correlation and Partial Auto-correlation Plots
fig = plt.figure(figsize=(12, 8))
ax1 = fig.add_subplot(211)
fig = sm.graphics.tsa.plot_acf(data_train_df['Close'], lags=160, ax=ax1)
ax2 = fig.add_subplot(212)
fig = sm.graphics.tsa.plot_pacf(data_train_df['Close'], lags=20, ax=ax2)
plt.show()
```

```python
#ARIMA
# Train the model and get predictions

# Step 7: Fit ARIMA model on training data
model = ARIMA(data_train_df['Close'], order=(2, 1, 0)).fit()

# Step 8: Make predictions on test data
history = list(data_train_df['Close'])
predictions = []
length = len(data_test_df)
for t in range(length):
    model_fit = ARIMA(history, order=(2, 1, 0)).fit()
    yhat = model_fit.forecast()[0]
    predictions.append(yhat)
    history.append(data_test_df.iloc[t]['Close'])

# Step 9: Save and plot predictions
predictions_df = pd.DataFrame({'Date': data_test_df.index, 'Close': predictions}).set_index('Date')
predictions_df.to_csv('./ARIMA_predictions.csv')
```

```python
test_set = data_test_df
training_set = data_train_df

#ARIMA
# Plot ARIMA results

make_plot(title='ARIMA: Stock Price Prediction', xlabel='Time', ylabel='Close Price')
plt.plot(test_set['Close'], label='Actual Stock Price')
plt.plot(predictions_df['Close'], label='Predicted Stock Price')
plt.show()

# Step 10: Residual analysis
residuals = pd.DataFrame(model.resid)
fig, ax = plt.subplots(1, 2, figsize=(12, 6))
residuals.plot(title='Residuals', ax=ax[0])
residuals.plot(kind='kde', title='Residual Density', ax=ax[1])
plt.show()
residuals.to_csv('./ARIMA_residuals.csv')

# Step 11: Evaluation metrics and diagnostics
MSE = metrics.mean_squared_error(test_set['Close'], predictions)
RMSE = MSE**0.5
MAE = metrics.mean_absolute_error(test_set['Close'],predictions)
R2 = metrics.r2_score(test_set['Close'],predictions)
print('MSE: %.5f' % MSE)
print('RMSE: %.5f' % RMSE)
print('MAE: %.5f' % MAE)
print('R2: %.5f' % R2)

adf_test(training_set['Close'].values)
adf_test(training_set['diff_1'].dropna().values)

# Step 12: Plot difference fitting
predicted_diff = pd.Series(model.fittedvalues, index=training_set.index[1:])
make_plot(title='Fitted vs Actual Differences', xlabel='Time', ylabel='First Difference')
plt.plot(training_set['diff_1'], label='Actual First Difference')
plt.plot(predicted_diff, label='Fitted First Difference')
plt.show()
```
