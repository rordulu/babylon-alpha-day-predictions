#### Inverse Scaling

One important thing to consider is the scale of the datasets, because the model was trained on scaled data that needs to be transformed back to its original price range.

Therefore, we have to load the same scaler used for transforming the original dataset and apply it to restore the prediction outputs to their actual distribution.

```python
# # Prepare scaler model name and location
# scaler_model_location = "..//models//"
# scaler_model_name = "google_stock_price_scaler"
# scaler_model_ext = "gz"

# # Store the scaler model
# sc = joblib.load(scaler_model_location + scaler_model_name + "." + scaler_model_ext)
```

The scaler was initially fitted on data with 6 columns encompassing all the features considered. However, during the inverse transformation process, our focus is solely on transforming the "Open" price.

```python
# IF IT WAS between 0 and 1, make sure true predictions are the ones that are above 0.5
y_train_predict_inv = (y_train_predict_inv > 0.5).astype(float)
y_validate_predict_inv = (y_validate_predict_inv > 0.5).astype(float)
y_test_predict_inv = (y_test_predict_inv > 0.5).astype(float)


print(f"y_train_predict_inv: shape: {y_train_predict_inv.shape}, first 5: {y_train_predict_inv[:5]}")
print(f"y_validate_predict_inv: shape: {y_validate_predict_inv.shape}, first 5: {y_validate_predict_inv[:5]}")
print(f"y_test_predict_inv: shape: {y_test_predict_inv.shape}, first 5: {y_test_predict_inv[:5]}")
```

```python
recent_samples = 50
plt.figure(figsize=(18,6))

def assess_performance(real, predict, train_data=""):

    num_of_true_positives = 0
    num_of_false_positives = 0
    

    predict = predict.flatten()
    real = real.flatten()


    print(predict)
    print(real)
    for i, val in enumerate(predict):
        if val >= 1.00005 and real[i] >= 1.00005:
            num_of_true_positives = num_of_true_positives + 1
        if val >= 1.00005 and real[i] < 1.00005:
            num_of_false_positives = num_of_false_positives + 1
    
    
    # true_positives = predict.flatten() * real.flatten()
    # false_positives = predict.flatten() - real.flatten()

    # num_of_true_positives = np.sum(true_positives > 0)
    # num_of_false_positives = np.sum(false_positives > 0)

    if num_of_true_positives + num_of_false_positives > 0:
        success_ratio = num_of_true_positives / (num_of_true_positives + num_of_false_positives)
    else:
        success_ratio = 0

    print(f"num_of_true_positives for: {train_data}: {num_of_true_positives}")
    print(f"num_of_false_positives for: {train_data}: {num_of_false_positives}")
    print(f"success ratio for: {train_data}: {100 * success_ratio}%")


assess_performance(y_train_inv, y_train_predict_inv, "Train")
assess_performance(y_validate_inv, y_validate_predict_inv, "Validate")
assess_performance(y_test_inv, y_test_predict_inv, "Test")
```

Despite the exact value predictions might be slightly different from real prices, the strong performance of this model is primarily derived from its ability to consistently mirror the actual trends across almost all phases. This is the most significant takeaway of stock price predictions analysis.

```python
recent_samples = 50
plt.figure(figsize=(18,6))

def assess_performance(real, predict, train_data=""):
    
    true_positives = predict.flatten() * real.flatten()
    false_positives = predict.flatten() - real.flatten()

    num_of_true_positives = np.sum(true_positives > 0)
    num_of_false_positives = np.sum(false_positives > 0)

    success_ratio = num_of_true_positives / (num_of_true_positives + num_of_false_positives)

    print(f"num_of_true_positives for: {train_data}: {num_of_true_positives}")
    print(f"num_of_false_positives for: {train_data}: {num_of_false_positives}")
    print(f"success ratio for: {train_data}: {100 * success_ratio}%")


assess_performance(y_train_inv.flatten(), y_train_predict_inv.flatten(), "Train")
assess_performance(y_validate_inv.flatten(), y_validate_predict_inv.flatten(), "Validate")
assess_performance(y_test_inv.flatten(), y_test_predict_inv.flatten(), "Test")
    
train_diff = y_train_predict_inv.flatten() - y_train_inv.flatten()
validate_diff = y_validate_predict_inv.flatten() - y_validate_inv.flatten()
test_diff = y_test_predict_inv.flatten() - y_test_inv.flatten()



plt.plot(data_train_dates[-recent_samples:], train_diff[-recent_samples:,], label="Training Data Prediction Diff", color=train_actual_color, linewidth=2)

plt.plot(data_validate_dates, validate_diff, label="Validation Data Prediction Diff", color=validate_predicted_color, linewidth=2)

plt.plot(data_test_dates, test_diff, label="Testing Data Prediction Diff", color=test_predicted_color, linewidth=2)

plt.title("Difference between prediction and real value (prediction - real)")
plt.xlabel("Time")
plt.ylabel("Stock Price (USD)")
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
plt.gca().xaxis.set_major_locator(mdates.MonthLocator(interval=1))
# plt.xticks(rotation=45)
plt.legend()
plt.grid(color="lightgray")
```

```python
def plot_for_series(real, predicted, set_name=""):

    print(f"real: {real.shape}")
    print(f"predicted: {predicted.flatten().shape}")
    
    real_diff = np.diff(real)
    predicted_diff = np.diff(predicted.flatten())

    whether_pos_or_neg = real_diff / predicted_diff

    wrong_direction_guesses = np.sum(whether_pos_or_neg < 0)
    total_length = len(whether_pos_or_neg)
    print (f"wrong direction guess for {set_name}: {100 * wrong_direction_guesses / total_length}%")

    

plot_for_series(y_train_inv, y_train_predict_inv, "Training")
plot_for_series(y_validate_inv, y_validate_predict_inv, "Validation")
plot_for_series(y_test_inv, y_test_predict_inv, "Test")
```
