```python
import pandas as pd
import xgboost as xgb
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix

%run "../helpers/data_manipulator.py"
```

```python
import numpy as np
import random
import os

np.random.seed(42)
random.seed(42)
os.environ["PYTHONHASHSEED"] = "42"
```

```python
add_prev_day = False

do_binary_reg = True
use_log_for_linear_reg = False
use_horizon = True

all_data = pd.DataFrame()
all_data_val = pd.DataFrame()

N1 = 18
N2 = 20
gamma = 0.8


for ticker in my_stocks:
    # GET RAW DATA FOR THE TICKER
    stock_file, earnings_file = fetch_finance_data_for_tickers(ticker, "10y", refresh=False, verbose=False, incl_earnings=True)
    data = pd.read_csv(stock_file)
    earnings = pd.read_csv(earnings_file)
    data = data.drop(data[data['Volume'] == 0].index)
    data.reset_index(drop=True, inplace=True)
    data = make_new_features_for(data, earnings=earnings, use_adj_close=True)

    
    data, data_val, _ = split_data(data,
                                train_cut_date = "2024-03-01",
                                validate_cut_date = "2025-03-01",
                                test_end_date = "2025-03-10",
                                train_start_date = "2016-07-01"
                               )
    normalized_data = normalize_data(data, ticker)
    normalized_data_val = normalize_data(data_val, ticker)
    normalized_data["Close"] = data["Adj Close"]
    normalized_data_val["Close"] = data_val["Adj Close"]
    all_data = normalized_data
    all_data_val = normalized_data_val
    # all_data = pd.concat([all_data, data], ignore_index=True)
    # all_data_val = pd.concat([all_data_val, data_val], ignore_index=True)


    # 1. Prepare data
    features = lstm_features_v2 
        
    print("Maximum values:\n", all_data[features].iloc[0:].max().to_dict())
    print("\nMinimum values:\n", all_data[features].iloc[0:].min().to_dict())
    
        
    unique_tickers = all_data["TICKER"].unique()
    
    # Convert to a 2D array where rows = tickers, cols = Close values
    all_data_np = np.array([all_data[all_data["TICKER"] == ticker]["Close"].values for ticker in unique_tickers], dtype=object)
    all_data_np_val = np.array([all_data_val[all_data_val["TICKER"] == ticker]["Close"].values for ticker in unique_tickers], dtype=object)
    
    print("Maximum values:\n", all_data[features].iloc[0:].max().to_dict())
    print("\nMinimum values:\n", all_data[features].iloc[0:].min().to_dict())
    
    
    
    if add_prev_day:
        
        print(f"all_data: {all_data}")
        prev_day_df = all_data.shift(1).add_prefix("prev_day_")
        all_data = pd.concat([all_data, prev_day_df], axis=1)
        all_data = all_data.iloc[1:]
        # all_data = all_data.dropna()
    
        print(f"all_data: {all_data}")
    
    
        prev_day_df_val = all_data_val.shift(1).add_prefix("prev_day_")
        all_data_val = pd.concat([all_data_val, prev_day_df_val], axis=1)
        # all_data_val = all_data_val.dropna()
        all_data_val = all_data_val.iloc[1:]
    
    import pandas as pd
    import xgboost as xgb
    from xgboost import XGBRegressor
    from sklearn.model_selection import train_test_split
    from sklearn.metrics import mean_squared_error, r2_score
    from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
    from xgboost import plot_importance
    import matplotlib.pyplot as plt
    
    if use_horizon:
        # # Calculate 10-day moving average
        # all_data['ma_10'] = all_data['Close'].rolling(window=20).mean()
        
        # # Shift it -10 days to get future 10D MA
        # all_data['ma_10_future'] = all_data['ma_10'].shift(-20)
        
        # # Binary target: 1 if future MA is higher than current MA
        # all_data['target'] = (all_data['ma_10_future'] > all_data['Close']).astype(int)
        
        # Define symmetric weights centered at N2
        weights = {
            i: gamma ** abs(i - N2)  # weight decreases as distance from N2 increases
            for i in range(N1, N2 + 1)
        }
        
        # Create numerator using the symmetric weights
        numerator = sum([
            all_data['Close'].shift(-i) * weight
            for i, weight in weights.items()
        ])
        
        # Denominator is the sum of all weights
        denominator = sum(weights.values())
        
        # Calculate the future weighted average
        future_weighted_avg = numerator / denominator

        
        # Set as regression target
        all_data['future_avg_close'] = future_weighted_avg
        all_data['forward_log_return'] = all_data['future_avg_close'] / all_data['Close']
        
        # Drop NaNs from shifting
        all_data = all_data.dropna(subset=['forward_log_return'])
        
        if use_log_for_linear_reg:
            all_data['forward_log_return'] = np.log(all_data['forward_log_return'])
        
        # Binary classification target (optional)
        all_data['target'] = ((all_data['future_avg_close'] / all_data['Close']) > 1.0).astype(int)
    
    else:
        all_data['target'] = (all_data["Next-Day-Close-To-Todays-Close-Ratio"] > 0)
        all_data["forward_log_return"] = all_data['Next-Day-Close-To-Todays-Close-Ratio']
    
    # Drop NaNs caused by shifting
    
    
    if add_prev_day:
        prev_day_features = [f"prev_day_{col}" for col in features]
        X = all_data[features + prev_day_features]
        X_val = all_data_val[features + prev_day_features]  # same features used in training
    else:
        X = all_data[features]
        X_val = all_data_val[features]
    
    
    if do_binary_reg:
        y = all_data['target']
    else:
        y = all_data['forward_log_return']
    
    # 2. Train-test split (time-aware if it's time series!)
    X_train, X_test, y_train, y_test = train_test_split(X, y, shuffle=False, test_size=0.2)
    
    if do_binary_reg:
        print(f"doing binary classification")
    
        class_counts = y_train.value_counts()
    
        # Get number of negative and positive samples
        num_neg = class_counts[0]
        num_pos = class_counts[1]

        print(f"num_neg: {num_neg}, num_pos: {num_pos}")
        
        # Calculate scale_pos_weight
        scale_pos_weight = num_neg / num_pos
    
        # 3. Train XGBoost Regressor
        model = xgb.XGBClassifier(
            n_estimators=1024,
            max_depth=5,
            learning_rate=0.05,
            objective='binary:logistic',
            eval_metric='logloss',
            random_state=42,
            subsample=0.8,         # fraction of rows to sample per tree
            colsample_bytree=0.8,  # fraction of features per tree
            tree_method='hist',   # Optional: can improve speed + determinism
            enable_categorical=False,  # ensure consistent parsing
            # class_weight='balanced',
            scale_pos_weight=scale_pos_weight,
        )
    else:
        print(f"doing linear regression")
        # 3. Train XGBoost Regressor
        model = xgb.XGBRegressor(
            n_estimators=512,
            max_depth=8,
            learning_rate=0.01,
            objective='reg:squarederror',
            eval_metric='rmse'
        )
        
    model.fit(X_train, 
              y_train,
              # num_boost_round=500,
              eval_set=[(X_test, y_test)],
              # early_stopping_rounds=10,
              verbose=False
             )
    
    # 4. Evaluate regression performance


    folder = "..//xgboost-models//" + ticker + "//"
    # Save folder if not exists
    os.makedirs(folder, exist_ok=True)

    
    y_pred = model.predict(X_test)
    if do_binary_reg:
        # Compute metrics
        cm = confusion_matrix(y_test, y_pred)
        cr = classification_report(y_test, y_pred, output_dict=True)
        accuracy = accuracy_score(y_test, y_pred)
        
        # Print metrics
        print("Confusion Matrix:")
        print(confusion_matrix(y_test, y_pred))
        
        print("\nClassification Report:")
        print(classification_report(y_test, y_pred))
        
        print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
        
        # Create a more detailed metrics DataFrame
        metrics_df = pd.DataFrame([
            # Confusion Matrix Details
            {'Category': 'Confusion Matrix', 'Metric': 'True Negatives', 'Value': cm[0, 0]},
            {'Category': 'Confusion Matrix', 'Metric': 'False Positives', 'Value': cm[0, 1]},
            {'Category': 'Confusion Matrix', 'Metric': 'False Negatives', 'Value': cm[1, 0]},
            {'Category': 'Confusion Matrix', 'Metric': 'True Positives', 'Value': cm[1, 1]},
            
            # Classification Report Metrics
            {'Category': 'Class 0', 'Metric': 'Precision', 'Value': cr['0']['precision']},
            {'Category': 'Class 0', 'Metric': 'Recall', 'Value': cr['0']['recall']},
            {'Category': 'Class 0', 'Metric': 'F1-Score', 'Value': cr['0']['f1-score']},
            {'Category': 'Class 0', 'Metric': 'Support', 'Value': cr['0']['support']},
            
            {'Category': 'Class 1', 'Metric': 'Precision', 'Value': cr['1']['precision']},
            {'Category': 'Class 1', 'Metric': 'Recall', 'Value': cr['1']['recall']},
            {'Category': 'Class 1', 'Metric': 'F1-Score', 'Value': cr['1']['f1-score']},
            {'Category': 'Class 1', 'Metric': 'Support', 'Value': cr['1']['support']},
            
            {'Category': 'Overall', 'Metric': 'Accuracy', 'Value': accuracy},
            {'Category': 'Overall', 'Metric': 'Macro Avg Precision', 'Value': cr['macro avg']['precision']},
            {'Category': 'Overall', 'Metric': 'Macro Avg Recall', 'Value': cr['macro avg']['recall']},
            {'Category': 'Overall', 'Metric': 'Macro Avg F1-Score', 'Value': cr['macro avg']['f1-score']},
            
            {'Category': 'Overall', 'Metric': 'Weighted Avg Precision', 'Value': cr['weighted avg']['precision']},
            {'Category': 'Overall', 'Metric': 'Weighted Avg Recall', 'Value': cr['weighted avg']['recall']},
            {'Category': 'Overall', 'Metric': 'Weighted Avg F1-Score', 'Value': cr['weighted avg']['f1-score']}
        ])
        
        # Save to CSV
        filename = f"../xgboost-models/{ticker}/model_metrics.csv"
        metrics_df.to_csv(filename, index=False)


    else:
        print(f"Mean Squared Error: {mean_squared_error(y_test, y_pred):.6f}")
        print(f"R² Score: {r2_score(y_test, y_pred):.4f}")
    
    # 5. Predict full signals
    all_data['prediction'] = model.predict(X)
    
    # 6. Convert predicted return ratio to buy/sell signal
    # Buy = 1 if expected return is positive (i.e., predicted ratio > 1)
    if do_binary_reg:
        all_data['signal'] = (all_data['prediction'] > 0.5).astype(int)
    else:
        all_data['signal'] = (all_data['prediction'] > 0).astype(int)
    
    # Optional: Check some predictions
    print(all_data[['prediction', 'signal']].tail())
    
    plot_importance(model)

    # Predict return ratio
    all_data_val['prediction'] = model.predict(X_val)
    
    # Generate signal: Buy if predicted return > 0
    if do_binary_reg:
        all_data_val['signal'] = (all_data_val['prediction'] > 0.5).astype(int)
    else:
        all_data_val['signal'] = (all_data_val['prediction'] > 0).astype(int)


    initial_cash = 10000
    cash = 0
    position = initial_cash / all_data_val["Close"].iloc[0]
    portfolio_value = []
    
    for i in range(1, len(all_data_val)):
        price_today = all_data_val.iloc[i]['Close']
        signal = all_data_val.iloc[i - 1]['signal']  # we act on yesterday's prediction
    
        # Buy signal (if not already in position)
        if signal == 1 and position == 0:
            position = cash / price_today
            cash = 0
    
        # Sell signal (if holding)
        elif signal == 0 and position > 0:
            cash = position * price_today
            position = 0
    
        # Track portfolio value
        value = cash + position * price_today
        portfolio_value.append(value)
    
    # Final value (if still holding)
    final_value = cash + position * all_data_val.iloc[-1]['Close']
    total_return = final_value - initial_cash
    
    buy_signals_count = (all_data_val['signal'] == 1).sum()
    total_rows = len(all_data_val)
    
    print(f"buy_signals_count: {buy_signals_count}")
    print(f"total_rows: {total_rows}")

    # Compute stats
    final_value = portfolio_value[-1]
    total_return = final_value - initial_cash
    return_percent = 100 * total_return / initial_cash
    invested_percent = 100 * buy_signals_count / total_rows
    stock_change_percent = 100 * ((all_data_val["Close"].iloc[-1] / all_data_val["Close"].iloc[0]) - 1.0)
    
    # Create DataFrame
    summary_df = pd.DataFrame({
        "Final Portfolio Value ($)": [round(final_value, 2)],
        "Total Return ($)": [round(total_return, 2)],
        "Return (%)": [round(return_percent, 2)],
        "Time Invested (%)": [round(invested_percent, 2)],
        "Stock Change (%)": [round(stock_change_percent, 2)],
    })
    
    
    # Save to CSV or Excel
    summary_df.to_csv(os.path.join(folder, "performance_summary.csv"), index=False)
    # summary_df.to_excel(os.path.join(folder, "performance_summary.xlsx"), index=False)  # optional Excel version
    
    print("Saved performance summary to:", os.path.join(folder, "performance_summary.csv"))


    import matplotlib.pyplot as plt
    import os
    
    os.makedirs(folder, exist_ok=True)

    
    plt.figure()
    plt.plot(portfolio_value)
    plt.title("Portfolio Value Over Time")
    plt.xlabel("Days")
    plt.ylabel("Value ($)")
    plt.savefig(folder + "-portfolio-value")
    plt.legend()
    plt.grid(True)
    plt.show()
    
    plt.figure()
    plt.plot(all_data_val["Close"])
    plt.title("Close Value Over Time")
    plt.xlabel("Days")
    plt.ylabel("Value ($)")
    plt.savefig(folder + "-close-over-time")
    plt.legend()
    plt.grid(True)
    plt.show()
    
    plt.figure()
    plt.plot(all_data_val["signal"])
    plt.title("Signal Over Time")
    plt.xlabel("Days")
    plt.ylabel("Value")
    plt.savefig(folder + "-signal-over-time")
    plt.legend()
    plt.grid(True)
    plt.show()

    model.save_model(folder + "xgb_model.json")
```
