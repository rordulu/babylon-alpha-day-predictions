```python
import os
import numpy as np
import pandas as pd
import xgboost as xgb
import matplotlib.pyplot as plt
import gym
import torch
from gym import spaces
import sys
sys.path.append('../helpers')
from tradingenv import TradingEnv

%run "../helpers/data_manipulator.py"

# Assuming the TradingEnv class is already defined elsewhere
# from trading_env import TradingEnv

def load_portfolio_models(stocks):
    """
    Load XGBoost models for given stocks
    
    :param stocks: List of stock tickers
    :return: Dictionary of loaded models
    """
    models = {}
    for ticker in stocks:
        model_path = f"../xgboost-models/{ticker}/xgb_model.json"
        if os.path.exists(model_path):
            model = xgb.XGBClassifier()
            model.load_model(model_path)
            models[ticker] = model
        else:
            print(f"Warning: No model found for {ticker}")
    return models

def prepare_data_for_trading_env(stocks):
    """
    Prepare data in the format expected by TradingEnv
    
    :param stocks: List of stock tickers
    :return: data and price_data suitable for TradingEnv
    """
    # Initialize storage for our data
    normalized_data_dict = {}
    price_data = np.zeros((len(stocks), 0))  # Will expand this as we process each stock
    max_length = 0
    
    # Load data for each stock
    for i, ticker in enumerate(stocks):
        # Fetch and prepare data (using imported helper functions)
        stock_file, earnings_file = fetch_finance_data_for_tickers(ticker, "10y", refresh=False, verbose=False, incl_earnings=True)
        data = pd.read_csv(stock_file)
        earnings = pd.read_csv(earnings_file)
        
        # Drop zero volume days
        data = data.drop(data[data['Volume'] == 0].index)
        data.reset_index(drop=True, inplace=True)
        
        # Feature engineering
        data = make_new_features_for(data, earnings=earnings, use_adj_close=True)
        
        # Split data (validation period)
        _, data_val, _ = split_data(data,
                                   train_cut_date="2024-03-01",
                                   validate_cut_date="2025-04-04",
                                   test_end_date="2025-03-10",
                                   train_start_date="2016-07-01")
        
        # Normalize data
        normalized_data_val = normalize_data(data_val, ticker)
        normalized_data_val["Close"] = data_val["Adj Close"]
        
        # Store data
        normalized_data_dict[ticker] = normalized_data_val
        
        # Track the maximum length
        if len(normalized_data_val) > max_length:
            max_length = len(normalized_data_val)
    
    # Now create the price_data array with the correct shape
    price_data = np.zeros((len(stocks), max_length))
    
    # Fill in the price data
    for i, ticker in enumerate(stocks):
        prices = normalized_data_dict[ticker]['Close'].values
        price_data[i, :len(prices)] = prices
    
    # Prepare the observation data for TradingEnv
    # For simplicity, we'll use a format that works with TradingEnv
    features = lstm_features_v2  # Assuming this is defined in your data_manipulator
    
    # For each time step, we need the features for all stocks
    data = []
    for t in range(max_length):
        time_step_data = []
        for i, ticker in enumerate(stocks):
            df = normalized_data_dict[ticker]
            if t < len(df):
                # Get features for this ticker at this time step
                ticker_features = df[features].iloc[t].values
                time_step_data.append(ticker_features)
            else:
                # Pad with zeros if this ticker doesn't have data for this time step
                ticker_features = np.zeros(len(features))
                time_step_data.append(ticker_features)
        data.append(np.array(time_step_data))
    
    # Convert to numpy array with shape suitable for TradingEnv
    data = np.array(data)
    
    return data, price_data, normalized_data_dict

def run_xgboost_with_trading_env(stocks, initial_cash=100000, benchmark_equal_weight=False):
    """
    Run XGBoost models with TradingEnv
    
    :param stocks: List of stock tickers
    :param initial_cash: Initial portfolio value
    :param benchmark_equal_weight: If True, use equal weight portfolio instead of model signals
    :return: Performance metrics
    """
    # Load models if not using equal weight
    models = {} if benchmark_equal_weight else load_portfolio_models(stocks)
    
    # Prepare data for TradingEnv
    data, price_data, normalized_data_dict = prepare_data_for_trading_env(stocks)
    
    # Create TradingEnv instance
    env = TradingEnv(data=data, price_data=price_data, tickers=stocks, initial_balance=initial_cash)
    env.divide_by_volatility = False
    env.single_stock_reward = False
    env.comission = 0.002
    env.buy_only_for_whats_sold = True
    # env.single_stock_portfolio_change = 0.1
    
    # Reset environment
    obs = env.reset()
    
    # Initialize tracking variables
    portfolio_value_history = [initial_cash]
    portfolio_allocation_history = {ticker: [0] for ticker in stocks}
    cash_balance_history = [initial_cash]
    action_history = []
    
    # Generate XGBoost signals for each time step
    all_signals = {}
    
    if not benchmark_equal_weight:
        for ticker in stocks:
            df = normalized_data_dict[ticker]
            features = lstm_features_v2  # Use the same features as in training
            
            if ticker in models:
                X = df[features]
                model = models[ticker]
                # Generate predictions
                df['prediction'] = model.predict(X)
                df['signal'] = (df['prediction'] > 0.5).astype(int)
                df['signal'] = df['signal'].rolling(window=3).apply(lambda x: 1 if (x == 1).all() else 0).fillna(0).astype(int)
                all_signals[ticker] = df['signal']
    else:
        for ticker in stocks:
            df = normalized_data_dict[ticker]
            features = lstm_features_v2  # Use the same features as in training
            
            if ticker in models:
                df['signal'] = 1
                all_signals[ticker] = df['signal']
        
    
    # Run the simulation
    done = False
    step = 0
    prev_action_values = []
    
    while not done and step < len(data) - 1:

        if benchmark_equal_weight:
            action_values = np.array([1.0/len(stocks)] * len(stocks) + [0.0], dtype=float)  # Equal allocation to all stocks
        else:
            action_values = np.zeros(len(stocks) + 1, dtype=float)  # +1 for cash position
            
            for i, ticker in enumerate(stocks):
                if ticker in all_signals and step < len(all_signals[ticker]):
                    if all_signals[ticker].iloc[step] == 1:
                        action_values[i] = 1.0
            
            # Normalize if any signals
            if np.sum(action_values) > 0:
                if np.sum(action_values) == 1:
                    action_values /= 2.0
                    action_values[-1] = 0.5
                action_values = action_values / np.sum(action_values)
            else:
                # If no buy signals, keep all in cash
                action_values[-1] = 1.0

        print(f"step: {step} and action_values: {action_values}")
        # Take step in environment
        obs, reward, done, info = env.step(action_values)
        prev_action_values = action_values
        
        # Record portfolio value and allocation
        total_value = env.cash_balance + sum(env.positions[i] * price_data[i, step] for i in range(len(stocks)))
        portfolio_value_history.append(total_value)
        cash_balance_history.append(env.cash_balance)
        
        # Record allocation percentages
        for i, ticker in enumerate(stocks):
            stock_value = env.positions[i] * price_data[i, step]
            if total_value > 0:
                allocation_pct = (stock_value / total_value) * 100
            else:
                allocation_pct = 0
            portfolio_allocation_history[ticker].append(allocation_pct)
        
        step += 1
    
    # Performance summary
    performance = {
        "Initial Cash": initial_cash,
        "Final Portfolio Value": portfolio_value_history[-1],
        "Final Cash Balance": cash_balance_history[-1],
        "Total Return ($)": portfolio_value_history[-1] - initial_cash,
        "Total Return (%)": 100 * (portfolio_value_history[-1] / initial_cash - 1),
        "Portfolio Value History": portfolio_value_history,
        "Cash Balance History": cash_balance_history,
        "Portfolio Allocation History": portfolio_allocation_history,
        "Action History": action_history
    }
    
    return performance

do_benchmark=True

model_results = run_xgboost_with_trading_env(my_stocks)

if do_benchmark:
    benchmark_results = run_xgboost_with_trading_env(my_stocks, benchmark_equal_weight=True)

# Print performance summary for both portfolios
print("Model-Based Portfolio Performance:")
for key, value in model_results.items():
    if key not in ['Portfolio Value History', 'Portfolio Allocation History', 'Action History']:
        print(f"{key}: {value}")

if do_benchmark:
    print("\nEqual-Weight Portfolio Performance:")
    for key, value in benchmark_results.items():
        if key not in ['Portfolio Value History', 'Portfolio Allocation History', 'Action History']:
            print(f"{key}: {value}")

# Plot portfolio values
plt.figure(figsize=(15, 15))

# Portfolio Value Subplot
plt.subplot(3, 1, 1)
plt.plot(model_results['Portfolio Value History'], label='Model-Based Portfolio', color='blue', linewidth=2)
if do_benchmark:
    plt.plot(benchmark_results['Portfolio Value History'], label='Equal-Weight Portfolio', color='red', linewidth=2)
plt.title('Portfolio Value Comparison')
plt.xlabel('Days')
plt.ylabel('Portfolio Value ($)')
plt.grid(True)
plt.legend()

# Cash Balance Subplot
plt.subplot(3, 1, 2)
plt.plot(model_results['Cash Balance History'], label='Model-Based Cash Balance', color='blue', linewidth=2)
if do_benchmark:
    plt.plot(benchmark_results['Cash Balance History'], label='Equal-Weight Cash Balance', color='red', linewidth=2)
plt.title('Cash Balance')
plt.xlabel('Days')
plt.ylabel('Cash Balance ($)')
plt.grid(True)
plt.legend()

# Portfolio Allocation Subplot
plt.subplot(3, 1, 3)
model_allocation = model_results['Portfolio Allocation History']
for ticker, allocations in model_allocation.items():
    plt.plot(allocations, label=f'Model-Based {ticker}', linewidth=2)

if do_benchmark:
    benchmark_allocation = benchmark_results['Portfolio Allocation History']
    for ticker, allocations in benchmark_allocation.items():
        plt.plot(allocations, label=f'Equal-Weight {ticker}', linestyle='--', linewidth=2)

plt.title('Portfolio Allocation Percentages')
plt.xlabel('Days')
plt.ylabel('Allocation (%)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
